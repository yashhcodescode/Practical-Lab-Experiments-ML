```python
import pandas as pd

# File name
CSV_PATH = "heart.csv"

# Read CSV file
df = pd.read_csv(CSV_PATH)

# 1. Display first 5 rows
print("FIRST 5 ROWS")
print(df.head())

# 2. Display dataset shape
print("\nDATASET SHAPE")
print(df.shape)

# 3. Display column names
print("\nCOLUMN NAMES")
print(df.columns)

# 4. Display data types
print("\nDATA TYPES")
print(df.dtypes)

# 5. Display basic information
print("\nDATASET INFORMATION")
df.info()

# 6. Display statistical summary
print("\nSTATISTICAL SUMMARY")
print(df.describe())

# 7. Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# 8. Check duplicate rows
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

# 9. Heart disease target distribution
print("\nHEART DISEASE DISTRIBUTION")
print(df["target"].value_counts())

# 10. Heart disease target percentage
print("\nHEART DISEASE PERCENTAGE")
print(df["target"].value_counts(normalize=True) * 100)
```
